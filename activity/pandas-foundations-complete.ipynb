## Setup

## Checkpoint 3: Build a mask

```python
recent_mask = talks["year"].between(2018, 2020)
```

```python
recent_mask.head(), recent_mask.shape, recent_mask.sum()
```

One `True` means that the corresponding talk occurred from 2018 through 2020.
The mask has 794 entries because it makes one decision for each row.

## Checkpoint 4: Filter and select together

```python
recent_talks = talks.loc[
    recent_mask,
    ["talk_name", "speaker", "conference", "year", "num_words"]
].copy()
```

```python
recent_talks.shape, recent_talks.head()
```

## Checkpoint 5: Sort the qualifying rows

```python
recent_talks.sort_values("num_words", ascending=False).head()
```

```python
recent_talks.sort_values(
    ["year", "num_words"],
    ascending=[False, False]
).head(10)
```

Year has priority in the second result. Every 2020 talk appears before every
2019 talk, even when a 2019 talk has a larger word count.

## Checkpoint 6: Create a variable

```python
recent_median = recent_talks["num_words"].median()

recent_talks["above_recent_median"] = (
    recent_talks["num_words"] > recent_median
)

recent_median
```

```python
recent_talks["above_recent_median"].value_counts()
```

## Final request

### 1. Build the row condition

```python
request_mask = (
    talks["year"].between(2019, 2020)
    & (talks["session"] == "sunday-morning")
    & talks["num_words"].between(1400, 1700)
)
```

### 2. Select rows and columns

```python
selected = talks.loc[
    request_mask,
    ["talk_name", "speaker", "conference", "year", "num_words"]
]
```

### 3. Apply the requested ordering

```python
result = selected.sort_values(
    ["year", "num_words"],
    ascending=[False, False]
)

result
```

### 4. Check the result

```python
checks = {
    "six_rows": len(result) == 6,
    "allowed_years": result["year"].isin([2019, 2020]).all(),
    "word_range": result["num_words"].between(1400, 1700).all(),
    "all_sunday_morning": talks.loc[result.index, "session"].eq(
        "sunday-morning"
    ).all(),
}

checks
```

The visible table places all 2020 talks above the 2019 talks. Within each year,
the word counts decrease from top to bottom.

## Reflection

The written request supplies three row conditions, five output columns, and two
sorting priorities. The code preserves that structure: the mask defines which
talks qualify, `.loc[]` retains the requested fields, and `sort_values()` puts
year ahead of word count when ordering the result.
